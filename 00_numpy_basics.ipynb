```python
import numpy as np
np.random.seed(42)
#Simulate 10 penalty kicks at a 75% conversion rate
kicks = np.random.random(10) #10 numbers between 1 and 10
scored = (kicks < 0.75) #True where number is below  0.75
goals = scored.sum() #count the Trues
print(f'Goals from 10 kicks: {goals}')
#Scale to 10000 kicks
big = np.random.random(10_000) #_ just makes the number readable
rate = (big < 0.75).mean()
print(f'Observed conversion rate {rate}')
print(f'Expected rate 0.7500')
```
