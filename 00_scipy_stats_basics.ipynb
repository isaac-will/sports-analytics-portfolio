```python
from scipy import stats
#Poisson distribution: average 1.4 goals per match
poisson = stats.poisson(mu=1.4)
print ('--- Probability of each score line ---')
for k in range(6):
    p = poisson.pmf(k)
    bar = '█' * int(p * 50) #contol + command + space and search for 'full block'
    print(f'P(goals = {k}):{p:.3f} {bar}')
print()
print('--- Quantiles (VAR instinct) ---')
print(f'5th pct   (VAR floor):   {poisson.ppf(0.05):.0f} goals')
print(f'25th pct:                {poisson.ppf(0.25):.0f} goals')
print(f'50th pct (median):       {poisson.ppf(0.50):.0f} goals')
print(f'75th pct:                {poisson.ppf(0.75):.0f} goals')
print(f'95th pct: (VAR ceiling): {poisson.ppf(0.95):.0f} goals') 
```
