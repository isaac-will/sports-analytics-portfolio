```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
goals = np.random.poisson(1.4,size=1000)
fig,ax = plt.subplots(nrows=1,ncols=1,figsize = (8,4))
ax.hist(goals, bins = range(0,9), edgecolor = 'white', color = '#534AB7', alpha = 0.9)
ax.set_xlabel('Goals scored')
ax.set_ylabel ('Number of simulated matches')
ax.set_title('Simulated EPL games - Poisson distribution (mean = 1.4)')
plt.tight_layout()
plt.savefig('first_chart.png', dpi=150)
plt.show()
# If an average happens at a fixed period of time, what is the probability that exactly k events will happen?
#If a team scores on average 1.4 goals per match (λ = 1.4), what is the probability they score exactly 2 goals?
```
