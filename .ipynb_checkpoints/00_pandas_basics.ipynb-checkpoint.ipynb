```python
import pandas as pd
```

```python
data = {'player': ['Salah','Haaland','Kane'],
        'goals': [19, 27, 21],
        'shots': [87, 112, 95],
        'minutes': [2850, 2700, 3060]}
df = pd.DataFrame (data)
#Normalize to 90 minutes
df['goals_p90'] = df['goals']/(df['minutes']/90)
df['shot_conv'] = df['goals']/df['shots']
print(df.sort_values('goals_p90', ascending = False).round(2))
```
